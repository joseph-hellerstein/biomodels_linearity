# CHANGE POINT ANALYSIS

```python
import matplotlib.pyplot as plt  # type: ignore
import numpy as np  # type: ignore
import pandas as pd  # type: ignore
from timecourse_iterator import TimecourseIterator # type: ignore
from typing import Optional, cast, List, Union

from src.system_discovery import SystemDiscovery
from src.piecewise_system_discovery import PiecewiseSystemDiscovery
from src.model import Model
from src.timecourse import Timecourse
from src.system_discovery_changepoint_detector import SystemDiscoveryChangepointDetector
```

# Helpers

```python
def makeDetector(model_num: Union[int, str] = 45, end_time: Optional[int] = None):
    if isinstance(model_num, int):
        model_name = Model.getBiomodelName(model_num)
    else:
        model_name = model_num
    tc = Timecourse.makeBiomodelDF(model_name, num_point=1000, end_time=end_time)
    disc = SystemDiscovery(
        tc.timecourse_df,
        threshold=0.001,
        alpha=0.001,
        is_normalize=True,
    )
    disc.fit()
    cpd = SystemDiscoveryChangepointDetector(disc)
    return SystemDiscoveryChangepointDetector(disc), tc
#
```

```python
def plotEvaluations(model_num: int, max_changepoint: int = 4, min_segment_length: int = 100, end_time: Optional[int] = None, is_plot: bool = True):
    detector, tc = makeDetector(model_num=model_num, end_time=end_time)
    detector.fit(max_changepoint=max_changepoint, min_segment_length=100, min_fractional_reduction=0.0)
    fig, axes = plt.subplots(2, 1, figsize=(15, 8))
    detector.plotTimecourseWithChangepoints(title=f"{model_num}", fig=fig, ax=axes[0])
    # Plot signal
    ax1 = axes[1]
    true_df, pred_df = detector._calculateNormalizedOneStepPredictions()
    diff_df = (true_df - pred_df).abs()
    signal_arr = diff_df.max(axis=1).to_numpy()
    ax1.plot(diff_df.index, signal_arr)
    #diff_df.plot(ax=ax1, title="|pred_deriv - true_deriv|")
    last_time = diff_df.index.to_numpy()[-1]
    ax1.set_xlim(0, last_time)
    ax1.set_title("Signal")
    plt.show
    if not is_plot:
        plt.close()
#
plotEvaluations(1045, max_changepoint=5, is_plot=False)
print("OK")
```

# Changepoints in specific models

```python
plotEvaluations(45, max_changepoint=10, end_time=62, min_segment_length=5)
```
